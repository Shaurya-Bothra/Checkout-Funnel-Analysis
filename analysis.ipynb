```python
import math
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

baseline = 0.484
relative_lift = 0.10
effect = proportion_effectsize(baseline * (1 + relative_lift), baseline)
n_per_arm = NormalIndPower().solve_power(effect_size=effect, alpha=0.05, power=0.8,
                                         alternative="two-sided")
print("Sessions needed per group:", math.ceil(n_per_arm))

sessions_reaching_stage = 3183
weekly = sessions_reaching_stage / 13
print("Weeks needed:", round(2 * n_per_arm / weekly, 1))
```
